# Im Browser öffnen: http://localhost:8080

### Projekt anlegen

- Einloggen

- Neues Projekt erstellen (Create oben rechts)

- Typ: Object Detection with Bounding Boxes

- Labels/Klassen definieren (z.B. „Gut“, „Defekt“, „Riss“...)

### Bilder importieren

- Im Menü „Daten“ oder „Import“ die 50 Bilder hochladen.

### Annotation starten

- Bild auswählen

- Bounding Box mit der Maus ziehen

- Label/Klasse auswählen

- Für jedes Bild alle relevanten Objekte markieren

### Export für YOLO (Ultralytics/YOLO11)
Im Projekt-Menü:
→ Exportieren
→ YOLO Format auswählen

Es werden pro Bild .txt-Dateien generiert (passend benannt).

### Kontrollieren
Prüft gemeinsam einige exportierte .txt-Files (Inhalt: Klassenindex, x, y, Breite, Höhe – jeweils relativ zum Bild).

Optional: Bilder & Label-Files in einen Ultralytics-YOLO-Ordnerstruktur verschieben:

    dataset/
    
      images/
    
        train/
    
        val/
    
      labels/
    
        train/
    
        val/

(Danach: Augmentierung erst nach dem Labeln, um Labels synchron zu den augmentierten Bildern zu erzeugen.)
